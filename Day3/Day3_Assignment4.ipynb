```python
# Q36.
# Count the frequency of each character in a string.
# Example
# Input:
text = "banana"

frequency = {}

for ch in text:
    if ch in frequency:
        frequency[ch] += 1
    else:
        frequency[ch] = 1

print(frequency)

```

```python
# Q37.
# Count the frequency of words in a sentence.
# Example
# Input
# I love Python and I love AI
sentence = "I love Python and I love AI"

words = sentence.split()
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
# Q39.
# Find duplicate values in a dictionary.
data = {
    "A": 10,
    "B": 20,
    "C": 10,
    "D": 30,
    "E": 20
}

count = {}

for value in data.values():
    if value in count:
        count[value] += 1
    else:
        count[value] = 1

for value in count:
    if count[value] > 1:
        print(value)
```

```python

# Q54.
# Merge dictionaries and, if duplicate keys exist, add their values.
# Example
d1 = {"A": 10, "B": 20}
d2 = {"B": 15, "C": 5}

result = d1.copy()

for key, value in d2.items():
    if key in result:
        result[key] += value
    else:
        result[key] = value

print(result)
```

```python
# Q55.
# Create a phonebook using dictionaries that allows the user to:
# •	Add a contact 
# •	Search for a contact 
# •	Update a contact 
# •	Delete a contact 
# •	Display all contacts 
# •	Exit the program 

phonebook= {}
print("Telephone directory Started !!!!!")
while True:
    print("\n--- PHONEBOOK ---")
    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Update Contact")
    print("4. Delete Contact")
    print("5. Display All Contacts")
    print("6. Exit")

    choice = input("please enter your choice: ")
    if choice == "1":
        name = input("Please enter name: ")
        number = input("please enter phn number: ")
        phonebook[name] = number
        print("Contact Saved Sucessfully!")
    elif choice == "2": 
        name = input("Enter name to search: ")
        
        if name in phonebook:
            print("Phone:", phonebook[name])
        else:
            print("Contact not found.")
    elif choice == "3" :
        name = input("Enter name to update: ")
        
        if name in phonebook:
            number = input("Enter number to update: ")
            phonebook[name] = number
            print("Phone Number updated Successfully: ", phonebook[name])
        else:
            print("Contact not found.")
    elif choice == "4" :
        name = input("Enter name yo want to delete: ")
        
        if name in phonebook:
            phonebook.pop(name)
            print("Contact deleted successfully.")
        else:
            print("Contact not found.")
    elif choice == "5" :
        
        for name in phonebook:
            print("Name: ",name," Phone: ", phonebook[name])
    elif choice == "6" :
        print("Closing")
        break
    else:
        print("Invalid Choice: ")
                    
    
```
