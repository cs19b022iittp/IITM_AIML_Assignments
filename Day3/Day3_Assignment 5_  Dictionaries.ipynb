```python
# 12. Student Attendance
```

```python
attendance = {
    "John": True,
    "Alice": False,
    "Bob": True,
    "David": None
}
```

```python
# Questions

# 1) Check whether Bob's attendance is available
if attendance.get("Bob") is not None:
    print("Bob's attendance is available")
else:
    print("Bob's attendance is not available")


# 2) Find students whose attendance has not been recorded
for student, status in attendance.items():
    if status is None:
        print("Attendance not recorded:", student)


# 3) Count present students
present = sum(1 for status in attendance.values() if status is True)
print("Present students:", present)


# 4) Count absent students
absent = sum(1 for status in attendance.values() if status is False)
print("Absent students:", absent)
```

```python
# 13. Movie Ratings
```

```python
ratings = {
    "Movie A": 4.5,
    "Movie B": None,
    "Movie C": 3.8,
    "Movie D": None
}
```

```python
# Questions

# 1) Check whether Movie C has a rating
if ratings.get("Movie C") is not None:
    print("Movie C has a rating")
else:
    print("Movie C has no rating")


# 2) Find all movies with no ratings
for movie, rating in ratings.items():
    if rating is None:
        print("No rating:", movie)


# 3) Count unrated movies
count = sum(1 for rating in ratings.values() if rating is None)
print("Unrated movies:", count)


# 4) Replace all missing ratings with 0
for movie in ratings:
    if ratings[movie] is None:
        ratings[movie] = 0

print(ratings)
```
