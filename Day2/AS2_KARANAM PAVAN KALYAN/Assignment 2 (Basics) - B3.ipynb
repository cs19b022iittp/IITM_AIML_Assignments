# Question 1  
## Create the following 3D array.

Layer 1

[1, 2, 3]
[4, 5, 6]

Layer 2

[7, 8, 9]
[10, 11, 12]

```python
# Write a Python program to:


import numpy as np
a = [[1,2,3],[4,5,6]]
b = [[7,8,9],[10,11,12]]
c= np.array([a,b])
# Print the complete 3D array.
print(c)
# Print the first layer.
print(c[0,::,::])
# Print the second row of the second layer.
print(c[1,::,::])
# Print the element 11.
print(c[1,1,1])
```

Expected Output

```python
b = c[::-1,::,::]
print("Layer2")
print(b[0])
print("Layer1")
print(b[1])
```

```python
# Question 8  

# Create the following nested dictionary.

students = {
    "Student1": {
        "Name": "John",
        "Age": 22
    },
    "Student2": {
        "Name": "Alice",
        "Age": 21
    },
    "Student3": {
        "Name": "David",
        "Age": 20
    }
}

# Write a program to

#     Print the name of Student2.
print(students["Student2"]["Name"])
#     Print the age of Student3.
print(students["Student3"]["Age"])

#     Display all student names using a loop.
for k in students.keys():
    print(k + " Name: " + students[k]["Name"])
```

```python
products = {}
def getdict(name,price,qnt):
    return  dict({"Product Name": name, "Price": price,"Quantity": qnt})
products["Laptop"] = getdict("Laptop",20000,2)
products["Mouse"] = getdict("Mouse",1000,2)
products["Keyboard"] = getdict("Keyboard",3000,2)
products["Monitor"] = getdict("Monitor",5000,2)

products["Printer"] = getdict("Printer",9000,1)
#    Display all products.
print("products:")
for v in products.values():
    print(v)
#   Print the product with the highest price.
hp = 0
costlyproduct = {}
for k in products.keys():
    if(hp < products[k]["Price"]):
        hp = products[k]["Price"]
        costlyproduct = products[k]
print("costly product:")
print(costlyproduct)
# Calculate the total inventory value
totalInventory = 0
for k in products.keys():
    totalInventory = totalInventory + products[k]["Price"]*products[k]["Quantity"]
print("Total inventory: " + str(totalInventory))
```

```python
stds = {"Alice":85,"Bob":92,"Charlie":78,"David":88,"Emma":95}
stds = dict(sorted(stds.items(), key=lambda item: item[1]))
names = list(stds)
print(stds)
avg = sum(stds.values()) / len(stds)
print("Highest: " + names[-1] + " " + str(stds[names[-1]]) )
print("Lowest: " +  names[0] + " " +  str(stds[names[0]]) )
print("Average: " + str(avg) )
```
